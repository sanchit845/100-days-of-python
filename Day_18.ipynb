# Numpy

```python
import numpy as np
print("numpy imported successfully")
```

```python
# to make an array - use np.array() function

a = np.array([[1,2,34],[1,2,34]])
print(a)
print(type(a))
```

```python
#Returns all values as true and 0 as false

n = np.array([1,2,3,4,0],dtype = bool)
print(n)
```

```python
# np.arange() - returns evenly spaced values within a given interval

print(np.arange(1,10))
print(np.arange(1,10,2))
print(np.arange(11))
```

```python
#np.reshape() - gives a new shape to an array without changing its data (rows,columns)

arr = np.arange(1,11)
np.reshape(arr,(5,2))
```

```python
# np.ones() and np.zeros() - returns a new array of given shape and type, filled with ones or zeros

print(np.ones((3,4),dtype = int))
print(np.zeros((3,4),dtype = int))
```

```python
# np.random() - returns an array of random numbers

np.random.random((3,4))
```

```python
# np.linspace() - returns evenly spaced numbers over a specified interval

np.linspace(-10,10,10)

#what this does is it returns 10 evenly spaced numbers from -10 to 10. The first two arguments are the start and end values, and the third argument is the number of evenly spaced samples to generate.
```

```python
# np.identity() - returns a square array with ones on the diagonal and zeros elsewhere

np.identity(4)
```

```python
a1 = np.arange(10)
a2 = np.arange(12,dtype = float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
```

## Features of an array

```python
# ndim - gives the number of dimensions of the array

print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
```

```python
# shape - gives the shape of the array

print(a1.shape)
print(a2.shape)
print(a3.shape)
```

```python
# size - gives the total number of elements in the array

print(a1.size)
print(a2.size)
print(a3.size)
```

```python
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

```python
# astype() - returns an array with the same data but a new data type

print(a1.dtype)
```

```python
a1 = a1.astype(np.int32)
print(a1.dtype)
```

```python
a = np.arange(1,13).reshape(3,4)
b = np.arange(13,25).reshape(3,4)

print(a)
print(b)
```

```python
# Scalar operations - operations that are performed on each element of the array

print(a)
print(a*2)
print(a + 5)

#Here any operation performed on an array is applied to each element of the array. This is called broadcasting.
```

```python
# Relational operators - returns a boolean array

print(b)
print(b > 18)
print(b <= 20)

# Here the boolean operators are applied to each element of the array and return a boolean array of the same shape as the original array.
```

```python
#Vector operations - operations that are performed on two arrays of the same shape

print(a,"\n\n",b)
```

```python
print(a + b)
print(a * b)

# Here the vector operations are applied to the corresponding elements of the two arrays and return a new array of the same shape as the original arrays.
```

```python
ran = np.random.random((3,3))
ran = np.round(ran*100)
ran
```

```python
# Some functions you can use in an array

print(np.max(ran))
print(np.min(ran))
print(np.sum(ran))
print(np.prod(ran))
```

```python
#we can use the axis parameter to specify the axis along which to perform the operation. For example, if we want to find the maximum value of each column, we can use axis=0. If we want to find the maximum value of each row, we can use axis=1.

print(np.max(ran,axis = 0)) # returns the maximum value of each column
print(np.max(ran,axis = 1)) # returns the maximum value of each row
```

```python
#Statistical functions - returns the mean, median, standard deviation and variance of the array

print(np.mean(ran)) #gives expected value of the array
print(np.median(ran)) #returns the median value of the array
print(np.std(ran)) # returns the standard deviation of the array
print(np.var(ran)) # returns the variance of the array
```

```python
# Dot product - returns the dot product of two arrays

c = np.arange(12).reshape(4,3)
print(np.dot(a,c)) # returns the dot product of a and b

# for the dot product to be defined, the number of columns in the first array must be equal to the number of rows in the second array. In this case, a has 4 columns and c has 4 rows, so the dot product is defined and returns a new array with shape (3,3).
```

```python
# round function is used to round off the value to the neaest integer.

roun = np.random.random((3,3))
print(roun,"\n")

roun = np.round(roun*100)
print(roun)
```

```python
#floor - throws the integer value of the number. It returns the largest integer less than or equal to the input value.
print(np.floor(roun))

# using floor - 5.9 will never go to 6, it will always go to 5.0. It is used to round down the value to the neareest integer.
```

```python
# ceil - throws the integer value of the number. It returns the smallest integer greater than or equal to the input value.
print(np.ceil(roun))

# using ceil - 2.1 will always round up to 3.0. It is used to round up the value to the nearest integer.
```

## Indexing and Slicing

### Stacking of arrays

```python
# Horizontal stacking - stacks arrays in sequence horizontally (column wise)

a = np.arange(1,13).reshape(3,4)
b = np.arange(13,25).reshape(3,4)
print(np.hstack((a,b)))
```

```python
# Vertical stacking - stacks arrays in sequence vertically (row wise)

print(np.vstack((a,b)))
```

### Splitting of array

```python
# Horizontal split - splits an array into multiple sub-arrays horizontally (column wise)

print(a2)
print(np.hsplit(a2,4)) # splits the array into 2 sub-arrays
```

```python
# Vertical split - splits an array into multiple sub-arrays vertically (row wise)

print(np.vsplit(a2,3)) # splits the array into 3 sub-arrays
```
