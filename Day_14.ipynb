# Inheritance 

In classes, If we want to make multiple classes with similar functionalities, there is one way to copy paste it.

But in this method, if we need to make an edit, we will have to do it with all the classes we made of that type which makes the time complexity very high and cost also very high. Other than this method, we have another method called - Inheritance.

In this method, the new classes inherits all the properties of the parent class, giving us space to modify it.

```python
# Single Inheritance: In single inheritance, a child class inherits from a single parent class. The child class can access the methods and attributes of the parent class.

class Employee:
    cls = 12
    def greet(self):
        print(f"Hello Sanchit!")

class Programmer(Employee):
    l="Python"
    def language(self):
        print(f"The language is {self.l}. The class is {self.cls}.") #Here, we are able to access the class variable of the parent class using self.cls.

a = Programmer()
a.greet() #Here, the greet function is inherited from the Employee class and we are able to use in the Programmer class.
a.language() #If I will do a =-Employee() then I will not be able to use the language function because it is not inherited from the Employee class.
```

In this way, the Programmer class inherited the properties and class attributes of Employee class

### Base class

The parent class

### Derived class 

The class that took the support of the base class to be made

```python
# Multiple Inheritance: In multiple inheritance, a child class can inherit from multiple parent classes. The child class can access the methods and attributes of all the parent classes.

class Employee:
    cls = 12
    def greet(self):
        print(f"Hello Sanchit!")

class coder:
    l="Python"
    def printlanguage(self):
        print(f"The language is {self.l}.")

class Programmer(Employee, coder):
    def language(self):
        print(f"The class is {self.cls}.") #Here, we are able to access the class variable of the parent class using self.cls.

a = Programmer()
a.greet() #Here, the greet function is inherited from the Employee class and we are able to use in the Programmer class.
a.language() #If I will do a =-Employee() then I will not be able to use the language function because it is not inherited from the Employee class.
a.printlanguage() #If I will do a =-coder() then I will not be able to use the printlanguage function because it is not inherited from the coder class.
```

```python
#This is called Multi-Class Inheritance, here. the Manager has the properties of the Programmer which has the properties of the Employee, and hence, the manager can access the properties of all.

class Employee():
    a = 1

class Programmer(Employee):
    b=2

class Manager(Programmer):
    c=3

call = Manager()
print(f" a is {call.a}, b is {call.b} and c is {call.c}")
```

```python
# super() is used to access the methods of a super class in the derived class 

class Employee():
    a = 1
    def __init__(self):
        print("This is a constructor of Employee")

class Programmer(Employee):
    b=2
    def __init__(self):
        super().__init__()
        print("This is a constructor of Programmer")

class Manager(Programmer):
    c=3
    def __init__(self):
        super().__init__()
        print("This is a constructor of Manager")

call = Manager()
print(f"a is {call.a}, b is {call.b} and c is {call.c}")

# When we call Manager, the constructor of Manager only works, but with super() tag, the constructor of the parent class also works.
```

### classmethod()

This is a method that is bound to the class and not to the object of the class

@classmethod is the decorator used.

```python
class Employee():
    a=1
    def show(self):
        print(a)

c=Employee
print(f"c.a before change is : {c.a}")
c.a=10
print(f"c.a after change is : {c.a}")

#Here we can see that the instance attribute overwrites the class attribute, to prevent this from happening, we use the @classmethod decorator.
```

```python
class Employee():
    a=1
    @classmethod
    def show(cls): #we use cls instead of self because we are using class method and not instance method.
        print(cls.a)

c=Employee()
print(f"c.a before change is : {c.a}")
c.a=10
c.show()

#Here we can see that the instance attribute overwrites the class attribute, to prevent this from happening, we use the @classmethod decorator.
```

### Property Decorators

Property : If there is a function, and I use the property decorator on it, it looks like a variable but contains a whole function in it.

Setter : The setter lets us define what should happen when someone assigns a value to the property.

```python
class Person:
    def __init__(self):
        self._age = 20

    @property
    def age(self):          # Getter/Property
        print("Getter called")
        return self._age

    @age.setter
    def age(self, value):   # Setter
        print("Setter called")
        self._age = value

Person1 = Person()
print(Person1.age)  # Calls the getter method
Person1.age = 25   # Calls the setter method
#print(Person1.age)  # Calls the getter method
```

### Operator Overloading



```python
class Number:
    def __init__(self,n):
        self.n = n

    def __add__(self, num):  #__add__ is a special method in Python that is used to define the behavior of the addition operator (+) for objects of a class. When you use the + operator between two objects, Python internally calls the __add__ method of the first object and passes the second object as an argument.
        return self.n + num.n

n=Number(5)
m=Number(10)

print(m+n)
```

```python
#Q1. 1. Create a class (2-D vector) and use it to create another class representing a 3-D vector.

class TwoDVector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def show(self):
        print(f"The vector is ({self.x}i, {self.y}j)")

class ThreeDVector(TwoDVector):
    def __init__(self, x, y, z):
        super().__init__(x, y)
        self.z = z

    def show(self):
        print(f"The vector is ({self.x}i, {self.y}j, {self.z}k)")

a = TwoDVector(1, 2)
a.show()
b = ThreeDVector(1, 2, 3)
b.show()
```

```python
#Q2. 2. Create a class 'Pets' from a class 'Animals' and further create a class 'Dog' from 'Pets'. Add a method 'bark' to class 'Dog'.

class Animal:
    pass

class Pets(Animal):
    pass

class Dog(Pets):
    @staticmethod
    def bark():
        print("Woof!")

a = Dog()
a.bark()  # Output: Woof!
```

```python
'''3. Create a class 'Employee' and add salary and increment properties to it.

Write a method 'salaryAfterIncrement' with a @property decorator and a setter which changes the value of increment based on the salary.'''

class Employee:
    salary = 50000
    increment = 1.5

    @property
    def salary_after_increment(self):
        return self.salary + (self.salary * (self.increment/100))

    @salary_after_increment.setter
    def salary_after_increment(self, salary):
        self.increment = ((salary/self.salary) -1)*100

e = Employee()
e.salary_after_increment=57500.000000000001
print(e.increment)  # Output: 57500.0
```

```python
#Q4. Write a class 'Complex' to represent complex numbers, along with overloaded operators '+' and '*' which add and multiply them.

class Complex():

    def __init__(self, r, i):
        self.r = r
        self.i = i

    def __add__(self,num):
        return Complex(self.r + num.r, self.i + num.i)

    def __mul__(self,num):
            return Complex(self.r * num.r - self.i * num.i, self.r * num.i + self.i * num.r)

    def __str__(self):
        return f"{self.r} + {self.i}i"

c = Complex(3, 4)
d = Complex(1, 2)
print(c+d)
print(c*d)
```

```python
#Q5. Write a class vector representing a vector of n dimensions. Overload the + and * operators which calculate the sum and the dot (.) product of them.

class vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, v):
        return vector(self.x + v.x, self.y + v.y)

    def __str__(self):
        return f"({self.x}i, {self.y}j)"

    def __mul__(self, v):
        return self.x * v.x + self.y * v.y

e = vector(1, 2)
f = vector(3, 4)
print(e+f)
print(e*f)
```
