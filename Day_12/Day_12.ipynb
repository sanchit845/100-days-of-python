# File I/O

We use files to read and write our code or output to save it in our Hard drive instead of the RAM.

### There are two types of files:

1. Text files : ( .txt ) ( .c)  used to store texts and information
2. Binary files : ( .jpg ), ( .dat ) etc.

```python
f = open("file.txt") #Opens a file
data = f.read() #read the contents of the file and store it in a variable
print(data)
f.close() #CLoses the file. This is important to free up system resources and avoid potential data loss.
```

Now by default, the file is opended in the read mode, we can also open the file in the write mode and edit the file.

1. r = read mode
2. w = write mode
3. a = appending
4. '+' = opening for updating
5. rb = read in binary mode
6. rt = open for read in text mode

If you open a file in write mode, and it does not exist, the file will be made and it will be opened in write mode.

```python
st = "This is another sample text file.\nIt contains multiple lines of text.\nThis is the third line."
f1 = open("file1.txt", "w") #Opens a file in write mode
f1.write(st) #Writes the string to the file
f1.close()
```

### Functions of file

1. readline() = reads one line from the file
2. readlines() = returns all the files in the file in the form of a list. 
3. with = with this function, we do not have to explicitly close the file.
 

```python
with open("file.txt") as f2: #Opens a file in read mode using a context manager
    print(f2.read()) #Prints the contents of the file

# here we don't have to explicitly close the file, as the context manager automatically takes care of it when the block is exited.
```

```python
#Q1. Write a program to read the text from a given file poems.txt and find out whether it contains the word twinkle.

with open("poems.txt") as f:
    content = f.read().lower() #Read the contents of the file
    if "twinkle" in content: #Check if the word "twinkle" is in the content
        print("The word 'twinkle' is present in the file.")
    else:
        print("The word 'twinkle' is not present in the file.")
```

```python
#Q2. The game() function in a program lets a user play a game and returns the score as an integer. You need to read a file Hi-score.txt which is either blank or contains the previous Hi-score. You need to write a program to update the Hi-score whenever the game() function breaks the Hi-score.

import random
def game():
    with open("Hi-score.txt", "r") as f:
        hiscore_val = int(f.read()) # Read the previous Hi-score from the file
    print("You are playing the game...")
    print(f"Previous Hi-score is: {hiscore_val}")
    score = random.randint(1, 100) # Simulating a score for the game
    print(f"Your score is: {score}")
    if (score > hiscore_val or hiscore_val == ""): 
            with open("Hi-score.txt", "w") as f:
                f.write(str(score)) # Update the Hi-score in the file 
    print(f"Hi-score is: {hiscore_val}")
    
game()
```

```python
#Q3. Write a program to generate multiplication tables from 2 to 20 and write them to different files. Place these files in a folder for a 13-year-old.

for i in range(2,21):
    for j in range(1,11):
        with open(f"Tables_Day_12_Q3/file{i}.txt", "a") as f:
            f.write(f"{i} x {j} = {i*j}\n")
```

```python
#Q4 A file contains a word Donkey multiple times. You need to write a program that replaces this word with #### by updating the same file.

with open("Q4.txt", "r") as f:
    text = f.read()

text = text.replace("Donkey", "####")

with open("Q4.txt", "a") as f:
    f.write(text)
```

```python
#Q5. Repeat program 4 for a list of such words to be censored.

words = ["The", "to", "a"]
with open("Q4.txt", "r") as f:
    text = f.read()

for word in words:
    text = text.replace(word, "####")

with open("Q5.txt", "a") as f:
    f.write(text)
```

```python
#Q6. Write a program to mine a log file and find out whether it contains python.

with open("Q6.txt", "r") as f:
    text = f.read().lower()

for word in text.split():
    if word == "python":
        print("python exists in this file")
        break
```

```python
#Q7. Write a program to find out the line number where python is present from Question 6.

with open("Q6.txt", "r") as f:
    text = f.readlines()

for line in text:
    if "python" in line.lower():
        print(f"Python exisits in line number: {text.index(line)+1}")
        print(f"The line goes like this: {line}")
        break
```

```python
#Q8. Write a program to make a copy of a text file this.txt.

with open("this.txt", "r") as f:
    text = f.read()

with open("copy.txt", "w") as f:
    f.write(text)
```

```python
#Q9. Write a program to find out whether a file is identical to and matches the content of another file.

with open("this.txt") as f:
    text1 = f.read()

with open("copy.txt") as f:
    text2 = f.read()

if text1 == text2:
    print("The files are identical.")
else:
    print("The files are not identical.")
```
