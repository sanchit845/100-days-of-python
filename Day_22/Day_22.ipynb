# Pandas

```python
import pandas as pd
import numpy as np
```

```python
check = ["Apple","Banana","Mango","Grapes","Orange","Pineapple","Strawberry","Watermelon"]
```

```python
# This is how we can create a series in pandas. A series is a one-dimensional array-like object that can hold any data type. It is similar to a list or an array in Python, but it has additional functionality and features.

pd.Series(check)
```

```python
nums = [ 13,21,45,51,24,514,21,54]
pd.Series(nums)
```

```python
# If we want to give a custom index to the series, we can do it like this:

marks = [88,91,99,90,85]
subjects = ["Maths","Science","SSt","English","Hindi"]
pd.Series(marks, index=subjects)
```

```python
# In order to give a name to the series, we can do it like this:

marks = pd.Series(marks,index = subjects, name = "Sanchit's marks")
print(marks)
```

```python
# We can create a pandas series from a dictionary as well. The keys of the dictionary will be the index of the series and the values will be the data of the series.

marks2 = {"Maths":88,"Science":91,"SSt":99,"English":90,"Hindi":85}
marks2_series = pd.Series(marks2, name = "Sanchit's marks")
print(marks2_series)
```

```python
# size - tell the number of elements in the series

marks2_series.size
```

```python
# dtype - tells the type of the data in the series

marks2_series.dtype
```

```python
# name - tells the name of the series

marks2_series.name
```

```python
# is_unique - tells whether the values in the series are unique or not

marks2_series.is_unique
```

```python
x = pd.Series([1,1,1,2,3,4,5])
x.is_unique

# This came out as False because the values in the series are not unique. There are multiple 1's in the series.
```

```python
# index - tells the index of the series

marks2_series.index
```

```python
# values - tells the values of the series

marks2_series.values
```

## Series using read_CSV

```python
# when tjere is a file with a single column, we can use squeeze() method to convert the dataframe into a series.

subs = pd.read_csv("subs.csv").squeeze("columns")
subs
```

```python
# when there are two columns in the file, we can use index_col parameter to set the index of the series.

score = pd.read_csv("kohli_ipl.csv", index_col="match_no").squeeze("columns")
print(score)
```

```python
# when there are multiple columns in the file, we can use squeeze() method to convert the dataframe into a series. But we need to specify the column name which we want to convert into a series.

pd.read_csv("bollywood.csv")
```

```python
movie = pd.read_csv("bollywood.csv", index_col = "movie").squeeze("columns")
movie
```

## Series Methods

```python
# head and tail - head is used to get the first n rows of the series and tail is used to get the last n rows of the series. By default, it will return 5 rows.

print(score.head(),"\n") #we can pass the number of rows we want to get as an argument to the head() and tail() methods
print(score.tail())
```

```python
# sample - sample is used to get a random sample of n rows from the series. By default, it will return 1 row.

print(movie.sample(),"\n") # default value of n is 1, so it will return 1 row if we don't pass any argument to the sample() method
```

```python
# value_counts - value_counts is used to get the count of unique values in the series. It will return a series with the unique values as the index and the count of each unique value as the values.

plot = movie.value_counts() # we can pass the parameter normalize = True to get the relative frequencies of the unique values in the series
```

```python
# sort_values() method is used to sort the values of the series in ascending or descending order. By default, it will sort the values in ascending order. We can pass the parameter ascending = False to sort the values in descending order.

score.sort_values(ascending=False) # this will give us the highest score of kohli in ipl
```

```python
# sort_values() method is used to sort the values of the series in ascending or descending order. By default, it will sort the values in ascending order. We can pass the parameter ascending = False to sort the values in descending order.

# write inplace = True to sort the values of the series in place. It will modify the original series and return None.
```

```python
# sort_index() method is used to sort the index of the series in ascending or descending order. By default, it will sort the index in ascending order. We can pass the parameter ascending = False to sort the index in descending order.
```

```python
# count() method is used to get the count of non-null values in the series. It will return an integer value.

score.count() # this will give us the number of matches played by kohli in ipl
```

```python
# mean, median, mode, std, var

print("Mean of the subs is: ",subs.mean())
print("Median of the subs is: ",subs.median())
print("Mode of the subs is: ",subs.mode())
print("Standard Deviation of the subs is: ",subs.std())
print("Variance of the subs is: ",subs.var())
```

```python
# min / max

print("The maximum value in the subs is: ",subs.max())
print("The minimum value in the subs is: ",subs.min())
```

```python
# describe - gives a summary of the series

print(subs.describe())
```
