# Python Virtual Environment

For a particular project, we don't generally use the global python environment to install our packages as it can overwrite some essential packages that may be used by other projects.

eg - I installed pandas 2.2 before, and then for a new project I need pandas 1.5 and I installed that, so in this case, 2.2 will be deleted and it will be overwritten by pandas 1.5

This will affect the working of the previous projects and the other corresponding projects. 

To prevent this, we create a virtual environment in python, whose scope is that particular project only.

### The package name to create a new virtual environment is virtualenv

command - pip install virtualenv

To create a new virtual environment - virtualenv _environment-name_

To activate the environment - .\(env name)\Scripts\activate.ps1



# Pip Freeze command

This command is used to look at the installed packages in our virtual environment.

To make it easier for recognition of pacakages, we use a file called requirements.txt to store and write down our packages, which we can use to install all the required packages instantly.

command to view installed packages - pip freeze

command to store the packages in a file - pip freeze > requirements.txt

command to install all packages at once - pip install -r requirements.txt
