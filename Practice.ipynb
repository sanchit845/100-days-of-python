1. Number Analyzer

Take an integer from the user and determine:

Whether it is positive, negative, or zero
Whether it is even or odd
Whether it is divisible by both 3 and 5

Print all applicable results.

```python
def p_n_z(num):
    if num == 0:
        print("Zero")
    elif num > 0:
        print("Positive")
    else:
        print("Negative")
def odd_even(num):
    if (num%2==0):
        print("Even")
    else:
        print("Odd")
def t_f(num):
    if (num%3==0 and num%5==0):
        print("The number is divisible by both 3 and 5")
    else:
        print("The number is not divisible by both 3 and 5")
num = int(input("Enter a number: "))
p_n_z(num)
odd_even(num)
t_f(num)
```

2. Student Result System

Take the marks of 5 subjects from the user.

Your program should:

Calculate total marks
Calculate percentage
Find the highest and lowest marks
Determine whether the student passed or failed
A student fails if they score below 33 in any subject

Try to store the marks in a list rather than creating five separate variables.

```python
l_Marks = []
for i in range(5):
    marks = int(input("Enter marks: "))
    l_Marks.append(marks)
l_Marks.sort()

sum = 0
for num in l_Marks:
    sum += num

print("Sum of the terms is", sum)

percentage = (sum/500)*100

print("The percentage is ",percentage)

print("The maximum marks is ", max(l_Marks))

print("The minimum marks is ", min(l_Marks))

if percentage <=33.3:
    print("The student has failed")
else:
    print("The student has passed")

```

3. Remove Duplicates — Without Losing Order

Given:

numbers = [4, 2, 7, 4, 2, 9, 7, 1, 9, 3]

Create a new list containing only the unique elements while preserving their original order.

Expected result:

[4, 2, 7, 9, 1, 3]

Restriction: Don't use set() directly to solve it.

```python
List = [4, 2, 7, 4, 2, 9, 7, 1, 9, 3]
new_list = []
for i in List:
    if i not in new_list:
        new_list.append(i)
print(new_list)
new_list.sort()
print("Sorted list is: ",new_list)
```

4. Word Frequency Counter

Take a sentence from the user.

For example:

"python is easy and python is powerful"

Create a dictionary containing how many times each word appears.

Expected:

{
    "python": 2,
    "is": 2,
    "easy": 1,
    "and": 1,
    "powerful": 1
}

Think about how you'll split the sentence and update the dictionary.

```python
dict = {}
sentence = input("Enter a sentence: ")
words = sentence.split(" ")
for word in words:
    if word in dict:
        dict[word] += 1
    else:
        dict[word] = 1
print(dict)
```

5. ATM Simulation

Create a simple ATM program.

The user starts with:

Balance = ₹10,000

Give them a menu:

1. Check Balance
2. Deposit
3. Withdraw
4. Exit

Requirements:

Deposit should increase the balance.
Withdrawal should only happen if sufficient money exists.
Invalid choices should be handled.
The menu should continue appearing until the user chooses Exit.

Think: Which loop would make the most sense here?

```python
balance = 10000
while True:
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")
    choice = int(input("Enter your choice: "))
    if choice == 1:
        print("Your balance is: ", balance)
    elif choice == 2:
        amount = int(input("Enter amount to deposit: "))
        balance += amount
        print("Your new balance is: ", balance)
    elif choice == 3:
        amount = int(input("Enter amount to withdraw: "))
        if amount > balance:
            print("Insufficient balance")
        else:
            balance -= amount
            print("Your new balance is: ", balance)
    elif choice == 4:
        break
    else:
        print("Invalid choice")
```
