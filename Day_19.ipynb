```python
##WHY IS NUMPY ARRAY PREFFERED OVER PYTHON LISTS

import numpy as np
import time

a = np.arange(10000000)
b = np.arange(10000000,20000000,dtype=np.int32)

start = time.time()
c = a + b
final = time.time() - start
print("Time taken by numpy array: ", final)
```

```python
a1 = [i for i in range(10000000)]
b1 = [i for i in range(10000000,20000000)]

c = []

start = time.time()
for i in range (len(a1)):
    c.append(a1[i] + b1[i])

final = time.time() - start

print("Time taken by python list: ", final)
```

## Boolean Indexing

## Broadcasting

It describes how numpy treats arrays with different shapes during arithmetic operations

Why do we need boradcasting??

When we need to perform operations on arrays with different sizes, the smaller array is "Broadcasted" across the larger array so that they have compatible shapes.

```python
n1 = np.array(np.random.randint(1,10,9).reshape(3,3))
n2 = np.array([1,2,3])
print(n1)
print(n2)

print(n1+n2) #this is called broadcasting. Here, n2 is broadcasted to the shape of n1 and then element-wise multiplication is performed.
```

```python
n3 = np.arange(12).reshape(3,4)
n4 = np.arange(12).reshape(4,3)

print(n3+n4) #this will give an error because the shapes of n3 and n4 are not compatible for broadcasting. The shapes must be compatible for broadcasting to work. The rule is that the dimensions must be either equal or one of them must be 1. In this case, the shapes are (3,4) and (4,3) which are not compatible for broadcasting.
```

## Handling missing values

```python
# In numpy, we use 'nan' to specify missing values. In python, we use 'None' to specify missing values. The difference is that 'nan' is a float value and 'None' is an object. This means that 'nan' can be used in mathematical operations, while 'None' cannot. For example, if we try to add 'nan' to a number, we will get 'nan' as the result. But if we try to add 'None' to a number, we will get an error.

n5 = np.array([1,2,3,4,np.nan,5,6,7])
n5
```

```python
print(np.isnan(n5))  #this will return a boolean array with True for the elements that are 'nan' and False for the elements that are not 'nan'
print(n5[np.isnan(n5)])  #this will return the elements that are 'nan' in the array n5
print(n5[~np.isnan(n5)])  #this will return the elements that are not 'nan' in the array n5
```
