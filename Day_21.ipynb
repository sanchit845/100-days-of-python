```python
def calculate(a,b,operation):
    if operation == "add":
        print( a + b)
    elif operation == "subtract":
        print( a - b)
    elif operation == "multiply":
        print( a * b)
    elif operation == "divide":
        print( a / b)
    else:
        print( "Invalid operation")

calculate(5,10,"add")
calculate(5,10,"subtract")
calculate(5,10,"multiply")
calculate(5,10,"divide")
calculate(5,10,"modulus")
```

```python
def greet_students(teacher_name, *student_names):
    print(f"Teacher: {teacher_name}")
    for student in student_names:
        print(f" - Student: {student}")

greet_students( "Dr. Verma", "Kartik", "Sanchit", "Rohan")
```

```python
int1 = input("Enter a number: ")
int2 = input("Enter another number: ")

try:
    result = int(int1) / int(int2)
    print(f"The result of {int1} divided by {int2} is: {result}")
    
except ValueError:
    print("Error: Invalid input! Please enter valid numbers.")
    
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

```python
numbers = [10, 45, 23, 67, 12, 89, 34, 89]

largest = float('-inf')  # Initialize largest to negative infinity
second_largest = float('-inf')  # Initialize second largest to negative infinity

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num !=largest:
        second_largest = num
print(second_largest)
```
