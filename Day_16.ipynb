### Walrus Operator

Assigns values to a variable as a part of an expression.

```python
if (n  := len([1,2,3,4,5])) >3:
    print(f"List has {n} elements, which is greater than 3.")

#Here we did not have to initialize the variable n before the if statement. The walrus operator (:=) allows us to assign a value to n while also checking the condition in the same line.
```

### Type hints

They are added in the code using a : "colon" and followed by -> for function return types.

```python
n : int = 3
name : str = "Sanchit"

def sum(a:int, b:int) -> int:
    return a + b

print(sum(2, 3))
print(type(n))
print(type(name))
print(type(sum(2,3)))
```

### Match case

It is similar to switch case as it is used to match values of a variable in multiple cases.

```python
def http_status(status):
    match status:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case 500:
            return "Internal Server Error"
        case _:
            return "Unknown Status"

print(http_status(200))
print(http_status(404))
print(http_status(500))
print(http_status(403))
```

### Exceptional Handling

We use try and except blocks to run our code inside those blocks.

By default, the try block runs, but if the try block fails, the code will jump into the except block rather than throwing an error.

We can also use an else block in the code, and it will be executed only if the try block is executed.

```python
try:
    a : int = "Sanchit"
    print(a + 5)

except Exception as e:
    print(e)

else:
    print("No exception occurred.") #This will be executed if no exception occurs in the try block.

# Here, we ran into the except block beacuse we tried to add an integer (5) to a string ("Sanchit"), which is not allowed in Python. This raises a TypeError, which is caught by the except block, and "ValueError" is printed.
```

We can also raise exceptions

```python
a = int(input("Enter a number: "))
b = int(input("Enter another number: "))

if (b == 0):
    raise ValueError("Cannot divide by zero.")
else:
    print(a / b)
```

## Enumerate

This function gives us a corresponding index to an iterable datatype. eg - if i call a list using enumerate function, it will give me the indexes along with the value.

```python
l = [53,523,4,33]

for index, item in enumerate(l):
    print(f"Index: {index} - Item: {item}")
```
