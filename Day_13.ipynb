# Object Oriented Programming

### Class

A class is a blueprint to make an object.

Once information is filled in a class, it becomes an object.

```python
class Employee:
    language = "Python" #this is a class attribute
    salary = 100000

sanchit = Employee()
sanchit.name = "Sanchit" #this is an instance attribute
print(sanchit.name, sanchit.language, sanchit.salary)
```

Here name is an instance attribute and language and salary are class attribute as they are inside the class.


Now, the instance attribute is prioritized over the class attribute.

```python
class Employee:
    language = "Python" #this is a class attribute
    salary = 100000

sanchit = Employee()
sanchit.language = "Javascript" #this is an instance attribute
print(sanchit.language, sanchit.salary)

#We passed the language attribute to the instance sanchit, so it will print Javascript instead of Python. But the class attribute is still Python. If we create another instance of Employee, it will still have the class attribute as Python.
```

### Static method

A method that only belongs to the class. It is used when the class is not taking any instance attribute as an arguement.

We make it by using the @staticmethod decorator

```python
class Employee:
    name = "Sanchit" #this is an instance attribute
    language = "Python" #this is a class attribute
    salary = 100000

    @staticmethod
    def get_info():
        print ("This is a static nethod line which does not take self as an argument. This is the language of the employee: Python")

sanchit = Employee()
sanchit.get_info()
```

### Dunder methods 
are those methods which are called by themselves. They are not needed to be called separately.

" __init__ "

```python
class Employee:
    language = "Python" #this is a class attribute
    saalary = 100000

    def __init__(self, name, language, salary):
        self.name = name
        self.language = language
        self.salary = salary
        print(f"The name of the employee is {name}. The language of the employee is {language}. The salary of the employee is {salary}.")

#Here we are not calling the def function, we are just creating an instance of the class Employee. When we create an instance of the class, the constructor is called automatically.

#Only init dunder method is called automatically when an instance of the class is created. Other methods are not called automatically.

sanchit = Employee("Sanchit", "Python", 100000)
print(sanchit.language, sanchit.saalary)
```

```python
'''Write a class Train that has methods to:

Book a ticket
Get status (number of seats)
Get fare information of a train running under Indian Railways
'''

class train():
    def __init__(self, name, fare, seats):
        self.name = name
        self.fare = fare
        self.seats = seats

    def get_status(self):
        print(f"Train name: {self.name}.")
        print(f"Seats available: {self.seats}")

    def fare_info(self):
        print(f"The fare of the train is: {self.fare}.")

    def book_ticket(self):
        if self.seats > 0:
            print(f"Your ticket has been booked. Your seat number is: {self.seats}.")
            self.seats -= 1
        else:
            print("Sorry, no seats available.")

tn = train("Rajdhani Express", 850, 3)

tn.get_status()
tn.fare_info()
tn.book_ticket()

```
